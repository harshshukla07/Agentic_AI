### Models Integration with OpenAI, Google Gemini and groq

```python
import os
from dotenv import load_dotenv
load_dotenv(override=True)

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
# This is the universal way for using any model just change the model name in init_chat_model function
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-4o-mini")
model
```

```python
## invoke model

response = model.invoke("hello, how are you?")
response
```

```python
response.content
```

### GROQ model integration

```python
import os
from langchain_groq import ChatGroq

model = ChatGroq(model = "llama-3.1-8b-instant")

response = model.invoke("Hey buddy, whats up?")
```

```python
response.content
```

## Streaming and Batch

### Streaming

```python
for chunk in model.stream("write me a 100 words paragraph on ai"):
    print(chunk.text, end="|", flush=True)
```

```python
for chunk in model.stream("why do peacock have such beautiful feathers?"):
    print(chunk.text, end="|", flush=True)
```

## Batch

```python
responses = model.batch([
    "how are you?",
    "What is the time in india right now?",
    "are you and AI and why"
])

for response in responses:
    print(response)
```

```python
responses = model.batch([
    "how are you?",
    "What is the time in india right now?",
    "are you and AI and why"
],
config={
    'max_concurrency':2, # Limit to 2 parallel calls
})

for response in responses:
    print(response)
```
